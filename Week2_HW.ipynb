```python
boston = pd.read_csv("/content/drive/MyDrive/Data 201/Datasets/Boston.csv")
boston.describe()
```

The dataset is 506 rows x 14 columns, with each row representing a suburb location in Boston, and each column representing one of the location's properties.

crim: per capita crime rate by town.

zn: proportion of residential land zoned for lots over 25,000 sq.ft.

indus: proportion of non-retail business acres per town.

chas: Charles River dummy variable (= 1 if tract bounds river; 0 otherwise).

nox: nitrogen oxides concentration (parts per 10 million).

rm: average number of rooms per dwelling.

age: proportion of owner-occupied units built prior to 1940.

dis: weighted mean of distances to five Boston employment centres.

rad: index of accessibility to radial highways.

tax: full-value property-tax rate per $10,000.

ptratio: pupil-teacher ratio by town.

lstat: lower status of the population (percent).

medv: median value of owner-occupied homes in $1000s.


```python
pd.plotting.scatter_matrix(boston[['crim', 'zn', 'indus', 'chas', 'nox', 'rm', 'age', 'dis', 'rad', 'tax', 'ptratio', 'lstat', 'medv']], figsize=(10, 11))
```

```python
boston[['crim']].median()
```

```python
boston[['ptratio']].mode()
```

As stated in the above paragraph, there are definitely some suburbs with particularly high per capita crime rates. The crim predictor has a range of (0.00632, 88.9762), while being overwhelmingly right skewed with a median of 0.25651, suggesting that the higher crime suburbs are potential outliers. Tax rates have a noticable gap in data, with a large number of suburbs having a signifcantly higher rate than the rest. The most common pupil-teacher ratio is 20.2, a value which stands out in an otherwise normal appearing bell curve.
